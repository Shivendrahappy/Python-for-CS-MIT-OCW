ALGORITHMS GUESS AND CHECK
CUBE ROOT

```python
cube = 8
for guess in range (abs(cube)+1):
  if guess**3 >= abs (cube):
    break

if guess **3 != abs (cube):
  print(cube,"is not a perfect cube")

else :
  if cube < 0:
    guess = -guess
  print("cube root of",cube,"is", guess)
```

```python
cube = 27
epsilon = 0.01
num_guesses = 0
low = 0
high = cube
guess = (high + low)/2.0
while abs(guess**3 - cube) >= epsilon:
  if guess**3 < cube :
    low = guess
  else:
    high = guess
  guess = (high + low)/2.0
  num_guesses += 1
print ("num_guesses =",num_guesses)
print (guess,"is close to the cube root of",cube)
```

Bisection method

```python
cube = 27
epsilon = 0.01
num_guesses = 0

low = 0
high = cube

guess = (high + low)/2.0
while abs(guess**3 - cube) >= epsilon:
  if guess**3 < cube :
    low = guess
  else:
    high = guess
  guess = (high + low)/2.0
  num_guesses += 1
print ("num_guesses =",num_guesses)
print (guess,"is close to the cube root of",cube)
```
